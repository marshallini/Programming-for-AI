```python
import xml.etree.ElementTree as ET
import pandas as pd

# Parse the XML file
tree = ET.parse('books.xml')
root = tree.getroot()
```

```python
books_list = []

for book in root.findall('book'):

    # Create a dictionary for the current book
    book_data = {}

    # Extract the 'id' from the book tag
    book_data['id'] = book.get('id')

    # Find and extract the text from the child tags
    book_data['title'] = book.find('title').text
    book_data['genre'] = book.find('genre').text
    book_data['published_year'] = int(book.find('published_year').text) #Convert year to integer

    # Find the 'author' tag and then find the nested name tags
    author = book.find('author')
    book_data['author_first_name'] = author.find('first_name').text
    book_data['author_last_name'] = author.find('last_name').text

    # Append the dictionary to our list
    books_list.append(book_data)

# Let's see the list of dictionaries we created
print(books_list)
```

```python
# Convert the list of dictionaries into a DataFrame
df = pd.DataFrame

# Display the final DataFrame
print(df)
```
